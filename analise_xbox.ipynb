```python
import pandas as pd
import numpy as np

# 1. Carregar os dados
df = pd.read_csv('xbox_game_pass.csv')

# 2. Visão geral do dataset
print("--- Primeiras Linhas ---")
display(df.head())

# 3. Métricas Principais
taxa_churn = (df['Status'].value_counts(normalize=True).get('Cancelado', 0)) * 100
media_horas = df['Horas_Jogadas_Mes'].mean()

print(f"\n📊 Taxa Geral de Churn: {taxa_churn:.2f}%")
print(f"⏱️ Média de Horas Jogadas/Mês: {media_horas:.1f} horas")

# 4. Taxa de Churn por Plano
churn_por_plano = df.groupby('Plano')['Status'].value_counts(normalize=True).unstack()['Cancelado'] * 100
print("\n🚨 Taxa de Churn por Plano (%):")
display(churn_por_plano.round(2))

# 5. Relação entre Horas Jogadas e Churn
horas_status = df.groupby('Status')['Horas_Jogadas_Mes'].mean()
print("\n⏱️ Média de Horas Jogadas por Status:")
display(horas_status.round(1))
```

```python
import plotly.express as px

# 1. Gráfico de Barras: Churn por Plano
fig_plano = px.bar(
    churn_por_plano.reset_index(),
    x='Plano',
    y='Cancelado',
    title='🚨 Taxa de Churn por Plano de Subscrição (%)',
    labels={'Cancelado': 'Taxa de Churn (%)', 'Plano': 'Plano Game Pass'},
    color='Cancelado',
    color_continuous_scale='Reds',
    text_auto='.1f'
)
fig_plano.show()

# 2. Boxplot: Distribuição de Horas Jogadas vs Status
fig_box = px.box(
    df,
    x='Status',
    y='Horas_Jogadas_Mes',
    color='Status',
    title='⏱️ Distribuição de Horas Jogadas por Status da Subscrição',
    labels={'Horas_Jogadas_Mes': 'Horas Jogadas / Mês', 'Status': 'Estado da Conta'},
    color_discrete_map={'Ativo': '#0e7a0d', 'Cancelado': '#d9534f'}
)
fig_box.show()
```
